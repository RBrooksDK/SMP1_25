# Exercise 1 - Discrete Random Variables
Binomial model for number of hits.

```python
from math import pow
from scipy.stats import binom

# Parameters
n = 100_000
p = 10_000 / pow(36, 6)

# Distribution
dist = binom(n=n, p=p)

# Results
p_zero_hits = dist.pmf(0)
mean = dist.mean()
var = dist.var()

print(f"n = {n}")
print(f"p = {p:.10f}")
print(f"P(X=0) = {p_zero_hits:.4f}")
print(f"mean = {mean:.4f}")
print(f"variance = {var:.4f}")
```
